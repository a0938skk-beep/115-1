```python
def total(p, q):
    return p * q

def discount(p, q):
    return total(p, q) * 0.9

p = int(input("請輸入單價："))
q = int(input("請輸入數量："))

print(f"原價金額：{total(p, q)}元")
print(f"9折金額：{int(discount(p, q))}元")
```
