```python
def calculate(price, quantity):
    subtotal = price * quantity      # 小計
    tax = subtotal * 0.05            # 稅金 = 小計 × 5%
    total = subtotal + tax           # 總金額
    return subtotal, tax, total      # 一次傳回多筆資料

price = int(input("請輸入單價："))
quantity = int(input("請輸入數量："))

subtotal, tax, total = calculate(price, quantity)

print("小計:", subtotal)
print("稅金:", tax)
print("總金額:", total)
```
